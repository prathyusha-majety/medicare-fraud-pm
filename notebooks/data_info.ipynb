```python
import pyarrow.parquet as pq
base = "/scratch/sm3664/medicare-fraud/processed"
files = {
    "Part B": f"{base}/part_b/labeled/part_b_labeled_2019_2023.parquet",
    "Part D": f"{base}/part_d/labeled/part_d_AE_labeled_2019_2023.parquet",
}
for name, f in files.items():
    s = pq.read_schema(f)
    print(f"\n=== {name}: {pq.ParquetFile(f).metadata.num_rows:,} rows x {len(s.names)} cols")
    for n, t in zip(s.names, s.types):
        print(f"{n}: {t}")
```
